# Clasificación de Perros y Gatos con Redes Neuronales Convolucionales (CNN)

En este cuaderno entrenaremos un modelo de Deep Learning capaz de distinguir entre imágenes de perros y gatos utilizando el dataset oficial de `tensorflow_datasets`. Aplicaremos técnicas modernas de preprocesamiento, aumento de datos (Data Augmentation) y exportaremos el modelo para su uso en una API.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import tensorflow as tf
import tensorflow_datasets as tfds
from keras import layers, models, callbacks

import os
import datetime
```

## 1. Carga del Dataset

Cargaremos el dataset `cats_vs_dogs` y lo dividiremos en conjuntos de entrenamiento, validación y test.

Para la carga del dataset, estamos utilizando la librería `tensorflow_datasets` (TFDS). Esta es una colección de datasets listos para usar con TensorFlow, JAX y otros frameworks de ML.

TFDS proporciona una interfaz sencilla para descargar y preparar datos. La función `tfds.load()` se encarga de todo el proceso:

*   **Descarga**: Si el dataset no está en tu máquina local, TFDS lo descarga automáticamente.
*   **Preparación**: Descomprime los archivos y los procesa en un formato estándar (generalmente `tf.data.Dataset`).
*   **Metadatos**: También proporciona metadatos útiles sobre el dataset, como el número de clases, la descripción de las características, etc.

En nuestro caso, `tfds.load('cats_vs_dogs', ...)` está cargando el dataset de imágenes de perros y gatos. El parámetro `as_supervised=True` asegura que los datos se entreguen en formato `(imagen, etiqueta)`.

Para más detalles, puedes consultar la documentación oficial de `tensorflow_datasets` aquí: [https://www.tensorflow.org/datasets/overview](https://www.tensorflow.org/datasets/overview)

El conjunto de datos `train_ds` (así como `test_ds`) es un `tf.data.Dataset` que, debido a que lo cargamos con `as_supervised=True`, produce elementos en el formato `(imagen, etiqueta)`.

*   **Imagen**: Es un tensor de TensorFlow que representa la imagen. Su forma será `(altura, anchura, canales)`, donde `canales` es 3 para imágenes RGB. Los valores de los píxeles estarán en el rango de 0 a 255 inicialmente, pero después de aplicar la capa `resize_and_rescale`, se redimensionarán a `IMG_SIZE` y se escalarán al rango de 0 a 1.

*   **Etiqueta**: Es un tensor escalar de TensorFlow que representa la clase de la imagen. Para el dataset `cats_vs_dogs`, los valores serán:
    *   `0` para gatos (cat)
    *   `1` para perros (dog)

Por lo tanto, al iterar sobre `train_ds`, cada elemento que se obtiene es una tupla `(image_tensor, label_tensor)`.

```python
(train_ds, val_ds, test_ds), metadata = tfds.load(
    'cats_vs_dogs',
    split=['train[:80%]', 'train[80%:90%]', 'train[90%:]'], # 80% train, 10% validation, 10% test
    with_info=True,
    as_supervised=True,
)
```

### Mostramos datos de prueba

Al mostrar los datos, podemos observar que las imágenes no tienen un tamaño uniforme (la altura y anchura varían entre ellas). Además, se trata de imágenes a color, lo cual se indica por la presencia de 3 canales (RGB).

```python
plt.figure(figsize=(10, 10))
for i, (image, label) in enumerate(train_ds.take(9)):  # Tomar 9 ejemplos del conjunto de entrenamiento
  ax = plt.subplot(3, 3, i + 1)
  plt.imshow(image.numpy().astype("uint8"))
  plt.title(metadata.features['label'].int2str(label)) # Mostrar la etiqueta (gato o perro)
plt.show()
```

## 2. Preparación de Datos

A diferencia de MNIST, las imágenes de este dataset no tienen el mismo tamaño, por lo que debemos redimensionarlas.

**¿Por qué no basta con usar una capa `layers.Resizing` en el bloque de preprocesamiento?**
Para entrenar, agrupamos las imágenes en lotes (**batches**). Si las fotos tienen tamaños distintos, TensorFlow no puede "apilarlas" en un bloque uniforme, lo que da un error de dimensiones antes siquiera de que el modelo empiece a procesarlas. Por eso, es obligatorio redimensionarlas **antes** de agruparlas con `.map()`.

**Entonces, ¿para qué sirve la capa `layers.Resizing`?**
Es muy útil cuando desplegamos el modelo (por ejemplo, en nuestra API). Al incluirla dentro del modelo, este se vuelve "autónomo": cualquier imagen que le enviemos, sea del tamaño que sea, será redimensionada automáticamente por la propia red antes de la primera convolución, sin que tengamos que programar ese ajuste fuera.

```python
IMG_SIZE = 180

def formatear_imagenes(image, label):
    # Redimensiona una sola imagen (necesario antes del batching)
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    return image, label

# Unifica el tamaño de todas las imágenes
train_ds = train_ds.map(formatear_imagenes)
val_ds = val_ds.map(formatear_imagenes)
test_ds = test_ds.map(formatear_imagenes)

# Procesa lotes (batch), mezcla y pre-carga para mayor velocidad
BATCH_SIZE = 32
train_ds = train_ds.shuffle(1000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
test_ds = test_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
```

### Mostramos los datos

El objetivo es comprobar que ha cambiado el tamaño

```python
plt.figure(figsize=(10, 10))

# Tomamos un solo lote del conjunto de entrenamiento
for images_batch, labels_batch in train_ds.take(1):
  for i in range(9): # Iteramos a través de las primeras 9 imágenes de ese lote
    ax = plt.subplot(3, 3, i + 1)
    # Accedemos a la imagen individual y su etiqueta dentro del lote
    plt.imshow(images_batch[i].numpy().astype("uint8"))
    plt.title(metadata.features['label'].int2str(labels_batch[i])) # Mostrar la etiqueta (gato o perro)
plt.show()
```

## 3. Bloque de Preprocesamiento

Utilizaremos las capas de Keras para unificar el tamaño de las imágenes y aplicar transformaciones aleatorias que ayuden a combatir el overfitting.

```python
IMG_SIZE = 180

# La capa de preprocesamiento realizará el cambio de tamaño y rescaling
capa_preprocesamiento = tf.keras.Sequential([
  layers.Resizing(IMG_SIZE, IMG_SIZE),
  layers.Rescaling(1./255)
])
```

## 4. Configuración de Tensoboard y Early Stopping

```python
root_logdir = os.path.join(os.getcwd(), "logs")

def get_run_logdir(model_name):
    run_id = datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
    return os.path.join(root_logdir, f"{model_name}_{run_id}")
```

```python
# Creamos el callback de EarlyStopping
early_stopping = callbacks.EarlyStopping(
    monitor='val_loss', # Métrica que queremos vigilar
    patience=5,        # Cuántas épocas esperar sin mejora antes de detenerse
    restore_best_weights=True # ¡Crítico! Nos quedamos con los pesos del punto óptimo
  )
```

## 5. Definición de los modelos

Diseñamos diferentes arquitecturas y posteriormente las compararemos.

### Modelo denso a color

```python
print("Entrenando Modelo denso color")
nombre_modelo = "Modelo_denso_color"
```

```python
# Creamos el callback para TensorBoard (irá registrando las salidas de nuestros modelos)
logdir = get_run_logdir(nombre_modelo)
tensorboard_cb = callbacks.TensorBoard(logdir)
```

```python
modelo_denso_color = models.Sequential([
  layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3)), # Capa de entrada explícita para el modelo (forma de una sola imagen)
  capa_preprocesamiento,

  layers.Flatten(),

  layers.Dense(256, activation='relu'),
  layers.Dense(128, activation='relu'),
  layers.Dense(64, activation='relu'),
  layers.Dense(1, activation='sigmoid') # Clasificación Binaria
], name=nombre_modelo)

modelo_denso_color.compile(optimizer='adam',
              loss=tf.keras.losses.BinaryCrossentropy(),
              metrics=['accuracy'])

modelo_denso_color.summary()
```

```python
epochs = 15
history = modelo_denso_color.fit(
  train_ds,
  validation_data=val_ds, # Usamos el dataset de validación directamente
  callbacks=[tensorboard_cb, early_stopping],
  epochs=epochs
)
```

### Modelo CNN a color sin data augmentation

```python
print("Entrenando Modelo CNN con data augmentation...")
nombre_modelo = "Modelo_color_con_data_aug"
```

```python
# Creamos el callback para TensorBoard (irá registrando las salidas de nuestros modelos)
logdir = get_run_logdir(nombre_modelo)
tensorboard_cb = callbacks.TensorBoard(logdir)
```

```python
modelo_data_aug_color = models.Sequential([
  layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3)), # Capa de entrada explícita para el modelo (forma de una sola imagen)

  capa_preprocesamiento, # layers.Rescaling(1./255)
  capa_data_augmentation, # Capas de aumento de datos

  layers.Conv2D(16, (3, 3), activation='relu'), # Ya no necesita input_shape aquí, lo infiere de Input layer
  layers.MaxPooling2D((2, 2)),

  layers.Conv2D(32, (3, 3), activation='relu'),
  layers.MaxPooling2D((2, 2)),

  layers.Conv2D(64, (3, 3), activation='relu'),
  layers.MaxPooling2D((2, 2)),

  layers.Flatten(),
  layers.Dense(128, activation='relu'),
  layers.Dense(1, activation='sigmoid') # Clasificación Binaria
], name=nombre_modelo)

modelo_data_aug_color.compile(optimizer='adam',
              loss=tf.keras.losses.BinaryCrossentropy(),
              metrics=['accuracy'])

modelo_data_aug_color.summary()
```

### Modelo CNN en Blanco y Negro

```python
print("Entrenando Modelo CNN en Blanco y Negro...")
nombre_modelo = "Modelo_CNN_BnW"

# Creamos el callback para TensorBoard
logdir = get_run_logdir(nombre_modelo)
tensorboard_cb = callbacks.TensorBoard(logdir)
```

#### Preparación de los datos en BN

```python
IMG_SIZE_BN = 180

def formatear_imagenes_bn(image, label):
    # Redimensiona una sola imagen a IMG_SIZE_BN
    image = tf.image.resize(image, (IMG_SIZE_BN, IMG_SIZE_BN))
    # Convierte la imagen a blanco y negro (1 canal)
    image = tf.image.rgb_to_grayscale(image)
    return image, label

# Cargamos el dataset nuevamente para aplicar el preprocesamiento en blanco y negro.
# Esto asegura que no usamos los datasets ya procesados a color y batcheados.
(train_ds_raw, val_ds_raw, test_ds_raw), metadata_bn = tfds.load(
    'cats_vs_dogs',
    split=['train[:80%]', 'train[80%:90%]', 'train[90%:]'], # 80% train, 10% validation, 10% test
    with_info=True,
    as_supervised=True,
)

# Unifica el tamaño y convierte a blanco y negro para todas las imágenes
# ¡Importante: aplicamos map a los datasets RAW, no a los ya batcheados!
train_ds_bn = train_ds_raw.map(formatear_imagenes_bn)
val_ds_bn = val_ds_raw.map(formatear_imagenes_bn)
test_ds_bn = test_ds_raw.map(formatear_imagenes_bn)

# Procesa lotes (batch), mezcla y pre-carga para mayor velocidad
BATCH_SIZE_BN = 32
train_ds_bn = train_ds_bn.shuffle(1000).batch(BATCH_SIZE_BN).prefetch(tf.data.AUTOTUNE)
val_ds_bn = val_ds_bn.batch(BATCH_SIZE_BN).prefetch(tf.data.AUTOTUNE)
test_ds_bn = test_ds_bn.batch(BATCH_SIZE_BN).prefetch(tf.data.AUTOTUNE)
```

#### Mostramos los datos

El objetivo es comprobar que ha cambiado el tamaño

```python
plt.figure(figsize=(10, 10))

# Tomamos un solo lote del conjunto de entrenamiento
for images_batch, labels_batch in train_ds_bn.take(1):
  for i in range(9): # Iteramos a través de las primeras 9 imágenes de ese lote
    ax = plt.subplot(3, 3, i + 1)
    # Accedemos a la imagen individual y su etiqueta dentro del lote
    img_to_display = images_batch[i].numpy().squeeze().astype("uint8")
    plt.imshow(img_to_display, cmap='gray')
    plt.title(metadata_bn.features['label'].int2str(labels_batch[i])) # Usamos metadata para las etiquetas
    plt.axis("off")
plt.show()
```

```python
print("Entrenando Modelo CNN en Blanco y Negro...")
nombre_modelo = "Modelo_CNN_BnW"

# Creamos el callback para TensorBoard
logdir = get_run_logdir(nombre_modelo)
tensorboard_cb_bn = callbacks.TensorBoard(logdir)

# Redefinimos el modelo para usar la capa de preprocesamiento BN
modelo_cnn_bn = models.Sequential([
  # La capa de entrada ahora espera 1 canal
  layers.Input(shape=(IMG_SIZE_BN, IMG_SIZE_BN, 1)),

  capa_preprocesamiento, # Usa la capa de preprocesamiento para BN

  layers.Conv2D(16, (3, 3), activation='relu'),
  layers.MaxPooling2D((2, 2)),

  layers.Conv2D(32, (3, 3), activation='relu'),
  layers.MaxPooling2D((2, 2)),

  layers.Conv2D(64, (3, 3), activation='relu'),
  layers.MaxPooling2D((2, 2)),

  layers.Flatten(),
  layers.Dense(128, activation='relu'),
  layers.Dense(1, activation='sigmoid') # Clasificación Binaria
], name=nombre_modelo)

modelo_cnn_bn.compile(optimizer='adam',
              loss=tf.keras.losses.BinaryCrossentropy(),
              metrics=['accuracy'])

modelo_cnn_bn.summary()

epochs = 15
history_cnn_bn = modelo_cnn_bn.fit(
  train_ds_bn, # Usamos el dataset en blanco y negro
  validation_data=val_ds_bn, # Usamos el dataset de validación en blanco y negro
  callbacks=[tensorboard_cb_bn, early_stopping],
  epochs=epochs
)
```

### Modelo CNN en Blanco y Negro con Dropout

```python
print("Entrenando Modelo CNN en Blanco y Negro con Dropout...")
nombre_modelo = "Modelo_CNN_BnW_dropout"

# Creamos el callback para TensorBoard
logdir = get_run_logdir(nombre_modelo)
tensorboard_cb = callbacks.TensorBoard(logdir)
```

```python
modelo_cnn_bn_dropout = models.Sequential([
  layers.Input(shape=(IMG_SIZE_BN, IMG_SIZE_BN, 1)), # Ahora la entrada es en blanco y negro

  capa_preprocesamiento,

  layers.Conv2D(16, (3, 3), activation='relu'),
  layers.MaxPooling2D((2, 2)),
  layers.Dropout(0.2), # Dropout añadido

  layers.Conv2D(32, (3, 3), activation='relu'),
  layers.MaxPooling2D((2, 2)),
  layers.Dropout(0.25), # Dropout añadido

  layers.Conv2D(64, (3, 3), activation='relu'),
  layers.MaxPooling2D((2, 2)),
  layers.Dropout(0.25), # Dropout añadido

  layers.Flatten(),
  layers.Dense(128, activation='relu'),
  layers.Dropout(0.5), # Dropout antes de la capa de salida (después de Flatten y Dense)
  layers.Dense(1, activation='sigmoid') # Clasificación Binaria
], name=nombre_modelo)

modelo_cnn_bn_dropout.compile(optimizer='adam',
              loss=tf.keras.losses.BinaryCrossentropy(),
              metrics=['accuracy'])

modelo_cnn_bn_dropout.summary()
```

```python
epochs = 15 # Puedes ajustar el número de épocas
history_cnn_bn = modelo_cnn_bn_dropout.fit(
  train_ds_bn,
  validation_data=val_ds_bn,
  callbacks=[tensorboard_cb, early_stopping],
  epochs=epochs
)
```

### Matrices de Confusión de los Modelos

Este código tiene como objetivo recopilar las etiquetas verdaderas (las clases reales a las que pertenecen las imágenes) de tus conjuntos de datos de prueba, `test_ds` (para imágenes a color) y `test_ds_bn` (para imágenes en blanco y negro).

Aquí te explico paso a paso lo que hace:

1.  **`y_true_color_list = []` y `y_true_bn_list = []`**: Se inicializan dos listas vacías. Estas listas almacenarán las etiquetas verdaderas de cada imagen para los modelos de color y los modelos en blanco y negro, respectivamente.

2.  **`for _, labels in test_ds.unbatch().as_numpy_iterator():`**:
    *   **`test_ds.unbatch()`**: Tus conjuntos de datos (como `train_ds`, `val_ds`, `test_ds`) están agrupados en lotes (`BATCH_SIZE=32`). `unbatch()` revierte esta operación, transformando el dataset para que produzca elementos individuales (una imagen y su etiqueta a la vez) en lugar de lotes de 32.
    *   **`.as_numpy_iterator()`**: Convierte el `tf.data.Dataset` en un iterador de Python que produce tensores de TensorFlow como objetos NumPy. Esto facilita el procesamiento fuera del grafo de TensorFlow.
    *   **`for _, labels in ...`**: Itera sobre este dataset. En cada iteración, se obtiene una imagen y su etiqueta. El guion bajo (`_`) se usa como un marcador para indicar que nos interesa la etiqueta (`labels`) pero no la imagen en sí en este punto. `labels` contendrá la etiqueta verdadera (0 para gato, 1 para perro) de la imagen actual.
    *   **`y_true_color_list.append(labels)`**: Cada etiqueta verdadera obtenida del dataset a color (`test_ds`) se añade a la lista `y_true_color_list`.

3.  **`y_true_color = np.array(y_true_color_list)`**: Una vez que todas las etiquetas han sido recopiladas en la lista, se convierten en un array de NumPy. Un array de NumPy es una estructura de datos muy eficiente para operaciones numéricas y es el formato que esperan muchas funciones de Scikit-learn (como `confusion_matrix`).

El mismo proceso se repite para `test_ds_bn` para obtener `y_true_bn`, que serán las etiquetas verdaderas para el modelo entrenado con imágenes en blanco y negro.

**¿Por qué es importante?**
Para calcular métricas de rendimiento como la matriz de confusión, necesitas dos cosas: las predicciones de tu modelo (`y_pred`) y las etiquetas verdaderas reales de los datos (`y_true`). Este bloque de código se encarga de preparar esas `y_true` de forma ordenada y en el formato correcto para su posterior uso.

```python
import numpy as np
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Recolectar etiquetas verdaderas del test_ds
y_true_color_list = []
for _, labels in test_ds.unbatch().as_numpy_iterator():
    y_true_color_list.append(labels)
y_true = np.array(y_true_color_list) # Usamos 'y_true' para todas las comparaciones

# Lista de modelos y su información correspondiente
model_infos = [
    {"name": "Modelo Denso Color", "model": modelo_denso_color, "test_ds": test_ds, "y_true": y_true},
    {"name": "Modelo CNN sin Data Aug", "model": modelo_basico_color, "test_ds": test_ds, "y_true": y_true},
    {"name": "Modelo CNN con Data Aug", "model": modelo_data_aug_color, "test_ds": test_ds, "y_true": y_true},
    {"name": "Modelo CNN BN sin Dropout", "model": modelo_cnn_bn, "test_ds": test_ds, "y_true": y_true},
    {"name": "Modelo CNN BN con Dropout", "model": modelo_cnn_bn_dropout, "test_ds": test_ds, "y_true": y_true}
]

# Definir etiquetas de clase (Cat=0, Dog=1)
class_labels = [metadata.features['label'].int2str(0), metadata.features['label'].int2str(1)] # 'cat', 'dog'

# Crear subplots para todas las matrices de confusión
# Se usarán 2 filas y 3 columnas, y se ocultarán los ejes no usados si hay menos de 6 modelos
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(20, 12))
axes = axes.flatten() # Aplanar el array de ejes para facilitar la iteración

for i, model_info in enumerate(model_infos):
    model = model_info["model"]
    test_ds_for_model = model_info["test_ds"]
    y_true_model = model_info["y_true"]
    model_name = model_info["name"]
    ax = axes[i]

    # Obtener predicciones (probabilidades y luego binarizarlas a 0 o 1)
    y_pred_prob = model.predict(test_ds_for_model, verbose=0)
    y_pred = (y_pred_prob > 0.5).astype(int) # Umbral de 0.5 para clasificación binaria

    # Calcular la matriz de confusión
    cm = confusion_matrix(y_true_model, y_pred)

    # Plotear la matriz de confusión
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=ax, cbar=False,
                xticklabels=class_labels,
                yticklabels=class_labels)

    ax.set_title(f'Matriz de Confusión: {model_name}')
    ax.set_xlabel('Predicción')
    ax.set_ylabel('Etiqueta Real')

# Ocultar cualquier subplot no utilizado
for j in range(len(model_infos), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
